```python
import jets
```

```python
config = {
    'exp_name': 'logtime',
    'visual': {
        'fields': ['q', 'v'],
        'spectra': ['z', 'e'],
        'means' : ['z_mean', 'e_mean'],
    },
    'atmosphere': {
        'n_grid': 256, 
        'params': {
            'initial': 'zero', 
            'forcing': 'norm',
            'p': 2,
            'nu': 4e-8, 
            'r': 0.01,
            'epsilon': 1.0,
            'kf': 16,
            'dkf': 2,
        }
    },
    'integrator': {
        'method': 'rk4',
        'interval': 1,
        'n_points': 100,
    },
}

model = jets.Driver()
model.setup(**config)
model.start()
print(model.descript())
```

```python
model.run(2, show=True, logtime=True, tau=2)
```
